```python
import pandas as pd
# Práctica Ventas de papelería
# Ambar Fierro 0060
```

```python
datos = {
    "Producto": ["Cuaderno", "Pluma", "Mochila", "Lápiz", "Regla"],
    "Precio": [15, 5, 500, 8, 20],
    "Cantidad": [1, 2, 1, 4, 2]
}

df = pd.DataFrame(datos)

df["Total"] = df["Precio"] * df["Cantidad"]

print("=== VENTAS DE PAPELERÍA ===")
print(df)
print("Venta total:", df["Total"].sum())
print("+-+-+-+-+jupyter Ambar Fierro +-+-+-+-+")
print("+-+-+-+-+- Ambar fierro 0060+-+-+-+--+")
```
